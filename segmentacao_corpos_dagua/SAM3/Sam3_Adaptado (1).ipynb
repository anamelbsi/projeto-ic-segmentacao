```python
#Bloco 02
import os
import cv2
import torch
import numpy as np

from PIL import Image

import matplotlib.pyplot as plt
```

```python
# Bloco 03 — Drive + Dataset

from google.colab import drive
import zipfile
import os

drive.mount('/content/drive')

ZIP_PATH = "/content/drive/MyDrive/waterbodiesdataset.zip"
EXTRACT_PATH = "/content/dataset"

if not os.path.exists(EXTRACT_PATH):
    with zipfile.ZipFile(ZIP_PATH, "r") as zip_ref:
        zip_ref.extractall(EXTRACT_PATH)

IMAGE_DIR = "/content/dataset/Water Bodies Dataset/Images"
MASK_DIR = "/content/dataset/Water Bodies Dataset/Masks"

print("Imagens:", IMAGE_DIR)
print("Máscaras:", MASK_DIR)

print("\nExiste Images?", os.path.exists(IMAGE_DIR))
print("Existe Masks?", os.path.exists(MASK_DIR))
```

```python
# Bloco 04 — Carregar imagem

image_files = sorted(os.listdir(IMAGE_DIR))

print("Total imagens:", len(image_files))

img_name = image_files[2]

print("Imagem escolhida:", img_name)

img_path = os.path.join(IMAGE_DIR, img_name)

image = Image.open(img_path).convert("RGB")

plt.figure(figsize=(8,8))
plt.imshow(image)
plt.title(img_name)
plt.axis("off")
plt.show()
```

```python
# Bloco 05 — Máscara real

mask_files = sorted(os.listdir(MASK_DIR))

mask_name = mask_files[2]

print("Máscara:", mask_name)

mask_path = os.path.join(
    MASK_DIR,
    mask_name
)

mask = Image.open(mask_path)

plt.figure(figsize=(8,8))

plt.imshow(mask, cmap="gray")

plt.title(mask_name)

plt.axis("off")

plt.show()
```

```python
# Bloco 06 — Conferir imagem e máscara

image_np = np.array(image)
mask_np = np.array(mask)

print("Imagem:", image_np.shape)
print("Máscara:", mask_np.shape)

print("Pixels da máscara:", np.unique(mask_np))
```
